```python
import matplotlib.pyplot as plt

# Group by industry and sum total layoffs
industry_layoffs = df.groupby('industry')['total_laid_off'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
industry_layoffs.plot(kind='bar', color='steelblue')
plt.title('Total Layoffs by Industry')
plt.xlabel('Industry')
plt.ylabel('Total Laid Off')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
