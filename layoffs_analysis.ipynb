```python
import pandas as pd

df = pd.read_csv('layoffs.csv')
df.head()
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt

# Group by industry and sum total layoffs
industry_layoffs = df.groupby('industry')['total_laid_off'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
industry_layoffs.plot(kind='bar', color='steelblue')
plt.title('Total Layoffs by Industry')
plt.xlabel('Industry')
plt.ylabel('Total Laid Off')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
top_companies = df.groupby('company')['total_laid_off'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))
top_companies.plot(kind='barh', color='darkorange')
plt.title('Top 10 Companies by Total Layoffs')
plt.xlabel('Total Laid Off')
plt.gca().invert_yaxis()  # highest at top
plt.tight_layout()
plt.show()
```
